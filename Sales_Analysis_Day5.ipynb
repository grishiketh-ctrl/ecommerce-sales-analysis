```python
import pandas as pd
import numpy as np

data = {
    "Date": [
        "2026-01-05", "2026-01-15", "2026-01-25",
        "2026-02-05", "2026-02-14", "2026-02-25",
        "2026-03-05", "2026-03-15", "2026-03-20",
        "2026-04-05", "2026-04-15", "2026-04-25",
        "2026-05-05", "2026-05-15", "2026-05-25",
        "2026-06-05", "2026-06-15", "2026-06-25",
        "2026-07-05", "2026-07-15", "2026-07-25",
        "2026-08-05", "2026-08-15", "2026-08-25"
    ],

    "Product": [
        "Laptop", "Phone", "Headphones",
        "Laptop", "Phone", "Tablet",
        "Laptop", "Phone", "Tablet",
        "Laptop", "Headphones", "Phone",
        "Laptop", "Phone", "Tablet",
        "Laptop", "Headphones", "Phone",
        "Laptop", "Phone", "Tablet",
        "Laptop", "Phone", "Headphones"
    ],

    "Region": [
        "South", "North", "East",
        "South", "North", "West",
        "East", "South", "North",
        "West", "East", "South",
        "North", "South", "East",
        "South", "West", "North",
        "East", "South", "West",
        "North", "East", "South"
    ],

    "Quantity": [
        2, 4, 8,
        3, 5, 3,
        4, 6, 4,
        3, 9, 5,
        5, 7, 4,
        6, 10, 8,
        7, 9, 5,
        8, 11, 7,
    ],

    "Price": [
        50000, 20000, 3000,
        50000, 20000, 25000,
        50000, 20000, 25000,
        50000, 3000, 20000,
        50000, 20000, 25000,
        50000, 3000, 20000,
        50000, 20000, 25000,
        50000, 20000, 3000
    ]
}

df = pd.DataFrame(data)

print(df)

df["Date"] = pd.to_datetime(df["Date"])

df["Revenue"] = df["Quantity"] * df["Price"]

print(df.head())
print("\nShape:", df.shape)
```

```python
df["Month"] = df["Date"].dt.month
df["Year_Month"] = df["Date"].dt.to_period("M")
print(df[["Date", "Month", "Year_Month"]].head())
```

```python
monthly_sales = df.groupby("Year_Month")["Revenue"].sum()

print(monthly_sales)
```

```python
monthly_quantity = df.groupby("Year_Month")["Quantity"].sum()

print(monthly_quantity)
```

```python
monthly_sales_growth = monthly_sales.pct_change() * 100

print(monthly_sales_growth)
```

```python
monthly_sales_ma = monthly_sales.rolling(window=2).mean()

print(monthly_sales_ma)
```

```python
trend_analysis = pd.DataFrame({
    "Sales": monthly_sales,
    "Sales_Growth_Percent": monthly_sales_growth,
    "Moving_Average": monthly_sales_ma
})

print(trend_analysis)
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(10, 5))

plt.plot(
    trend_analysis.index.astype(str),
    trend_analysis["Sales"],
    marker="o"
)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Revenue")

plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(10, 5))

plt.plot(
    trend_analysis.index.astype(str),
    trend_analysis["Sales"],
    marker="o",
    label="Actual Sales"
)

plt.plot(
    trend_analysis.index.astype(str),
    trend_analysis["Moving_Average"],
    marker="o",
    label="Moving Average"
)

plt.title("Sales Trend with Moving Average")
plt.xlabel("Month")
plt.ylabel("Revenue")

plt.xticks(rotation=45)
plt.legend()

plt.show()
```

```python
def test_no_missing_values(df):
    assert df.isnull().sum().sum() == 0
    print("✅ Missing value test passed")
```

```python
def test_no_negative_values(df):
    assert (df["Quantity"] >= 0).all()
    assert (df["Price"] >= 0).all()
    assert (df["Revenue"] >= 0).all()

    print("✅ Negative value test passed")
test_no_negative_values(df)
```

```python
def test_revenue_calculation(df):
    expected_revenue = df["Quantity"] * df["Price"]

    assert (df["Revenue"] == expected_revenue).all()

    print("✅ Revenue calculation test passed")
test_revenue_calculation(df)
```

```python
def test_trend_data(trend_analysis):
    assert len(trend_analysis) > 0
    assert "Sales" in trend_analysis.columns
    assert "Moving_Average" in trend_analysis.columns

    print("✅ Trend analysis test passed")
test_trend_data(trend_analysis)
```

```python
test_no_missing_values(df)
test_no_negative_values(df)
test_revenue_calculation(df)
test_trend_data(trend_analysis)
```
